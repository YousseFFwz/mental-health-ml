Data Preprocessing

```python
# Define target
TARGET = "Mental_Health_Risk"

# Remove identifier and target from features
X = df.drop(columns=["Student_ID", TARGET])

# Target variable
y = df[TARGET]


```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:")
print(y_train.value_counts())

print("\nTesting Target:")
print(y_test.value_counts())
```

```python
print("Train size:", len(X_train))
print("Test size:", len(X_test))

print("\nTrain distribution (%):")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTest distribution (%):")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
# Numerical Features
numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Categorical Features
categorical_features = X_train.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nNumerical Count:", len(numerical_features))
print("Categorical Count:", len(categorical_features))
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
numerical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Original Train Shape:", X_train.shape)
print("Processed Train Shape:", X_train_processed.shape)

print("\nOriginal Test Shape:", X_test.shape)
print("Processed Test Shape:", X_test_processed.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)
```

```python
# Train the complete pipeline

model_pipeline.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model_pipeline.predict(X_test)

print("Predictions:", y_pred[:10])
print("Actual Values:", y_test.iloc[:10].values)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model_pipeline.classes_
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model_pipeline.classes_
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

```python
rf_pipeline.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
y_pred_rf = rf_pipeline.predict(X_test)

print("First 10 predictions:", y_pred_rf[:10])
print("First 10 actual values:", y_test.iloc[:10].values)
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_rf = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", round(accuracy_rf, 4))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
import xgboost as xgb

print("XGBoost version:", xgb.__version__)
```

```python
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder

# Encodage de la variable cible
label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

# Création du pipeline
xgb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", XGBClassifier(
            n_estimators=100,
            max_depth=3,
            learning_rate=0.1,
            objective="multi:softprob",
            num_class=3,
            eval_metric="mlogloss",
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

```python
xgb_pipeline.fit(X_train, y_train_encoded)

print("XGBoost trained successfully!")
```

```python
y_pred_xgb = xgb_pipeline.predict(X_test)

print("Predictions:", y_pred_xgb[:10])
print("Actual values:", y_test_encoded[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_xgb = accuracy_score(y_test_encoded, y_pred_xgb)

print("XGBoost Accuracy:", round(accuracy_xgb, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test_encoded,
        y_pred_xgb,
        target_names=label_encoder.classes_
    )
)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring = {
    "accuracy": "accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro"
}

cv_results_lr = cross_validate(
    model_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)
```

```python
print("Accuracy per fold:", cv_results_lr["test_accuracy"])

print(
    "Mean Accuracy:",
    round(cv_results_lr["test_accuracy"].mean(), 4)
)

print(
    "Mean F1:",
    round(cv_results_lr["test_f1_macro"].mean(), 4)
)
```

```python
cv_results_rf = cross_validate(
    rf_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)

print("Accuracy per fold:", cv_results_rf["test_accuracy"])

print(
    "Mean Accuracy:",
    round(cv_results_rf["test_accuracy"].mean(), 4)
)

print(
    "Mean F1:",
    round(cv_results_rf["test_f1_macro"].mean(), 4)
)
```

```python
cv_results_xgb = cross_validate(
    xgb_pipeline,
    X_train,
    y_train_encoded,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)

print("Accuracy per fold:", cv_results_xgb["test_accuracy"])

print(
    "Mean Accuracy:",
    round(cv_results_xgb["test_accuracy"].mean(), 4)
)

print(
    "Mean F1:",
    round(cv_results_xgb["test_f1_macro"].mean(), 4)
)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Mean Accuracy": [
        cv_results_lr["test_accuracy"].mean(),
        cv_results_rf["test_accuracy"].mean(),
        cv_results_xgb["test_accuracy"].mean()
    ],
    "Mean F1": [
        cv_results_lr["test_f1_macro"].mean(),
        cv_results_rf["test_f1_macro"].mean(),
        cv_results_xgb["test_f1_macro"].mean()
    ]
})

comparison = comparison.sort_values(
    by="Mean F1",
).reset_index(drop=True)

comparison.round(4)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
param_grid = {
    "classifier__C": [0.01, 0.1, 1, 10, 100],
    "classifier__class_weight": [None, "balanced"],
    "classifier__solver": ["lbfgs"]
}
```

```python
tuning_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ]
)
```

```python
grid_search = GridSearchCV(
    estimator=tuning_pipeline,
    param_grid=param_grid,
    cv=cv,
    scoring="f1_macro",
    n_jobs=-1,
    verbose=1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross Validation Macro F1:")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:", accuracy_score(y_test, y_pred_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned))
```
